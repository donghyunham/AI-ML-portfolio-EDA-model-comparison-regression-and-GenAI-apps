```python
import pandas as pd
import numpy as np

data = {
    'Student_ID': range(1, 6),
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Math': [85, 92, np.nan, 78, 95],
    'English': [88, np.nan, 80, 92, 85],
    'Attendance': [95, 80, 100, 75, 90]
}
df = pd.DataFrame(data)

df['Math'] = df['Math'].fillna(df['Math'].mean())
df['English'] = df['English'].fillna(df['English'].mean())

df['Total_Score'] = df['Math'] + df['English']
df['Average'] = df['Total_Score'] / 2
df['Status'] = np.where((df['Average'] >= 85) & (df['Attendance'] >= 80), 'Pass', 'Fail')

summary = df.groupby('Status')['Average'].mean().reset_index()
display(summary)
```
